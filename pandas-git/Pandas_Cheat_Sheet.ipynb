```python

# Core imports
import pandas as pd
import numpy as np

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120)
```

## 📥 Reading Data

```python

# Read CSV
df = pd.read_csv("netflix_titles.csv")

# Peek at data
df.head()
```

## 🔍 Basic Inspection

```python

df.tail()
df.shape
df.columns
df.index
df.dtypes
df.info()
df.describe(include='all')
```

## 🧹 Data Cleaning

```python

# Check missing values
df.isna().sum()

# Fill missing values
df['director'] = df['director'].fillna("Unknown")

# Drop rows
df_dropped = df.dropna(subset=['date_added'])

# Drop columns
df_reduced = df.drop(columns=['show_id'])
```

## 🔄 Type Conversion

```python

df['release_year'] = df['release_year'].astype(int)
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
```

## 📌 Selection & Filtering

```python

# Column selection
df['title']
df[['title', 'type']]

# Row slicing
df.iloc[0:5]
df.loc[df['type'] == 'Movie']

# Conditional filtering
df[df['release_year'] > 2015]
```

## ➕ Feature Engineering

```python

# Create new column
df['title_length'] = df['title'].str.len()

# String operations
df['country'] = df['country'].str.upper()
```

## 📊 Sorting & Ranking

```python

df.sort_values(by='release_year', ascending=False)
df['release_year'].rank()
```

## 🧮 GroupBy & Aggregations

```python

df.groupby('type').size()

df.groupby('type').agg({
    'release_year': ['min', 'max', 'mean'],
    'title': 'count'
})
```

## 🔁 Apply, Map, Lambda

```python

df['is_recent'] = df['release_year'].apply(lambda x: x >= 2020)
```

## 🔗 Merge, Join, Concat

```python

df1 = df[['title', 'type']].head()
df2 = df[['title', 'release_year']].head()

pd.merge(df1, df2, on='title', how='inner')
pd.concat([df1, df2], axis=1)
```

## 🔄 Reshaping Data

```python

# Pivot table
pd.pivot_table(df, values='release_year', index='type', aggfunc='mean')

# Melt
pd.melt(df, id_vars=['title'], value_vars=['type', 'rating'])
```

## 🕒 Time Series

```python

df.set_index('date_added', inplace=True)
df.resample('Y').size()
df.reset_index(inplace=True)
```

## 📦 Duplicates

```python

df.duplicated().sum()
df.drop_duplicates()
```

## 🎲 Sampling

```python

df.sample(5, random_state=42)
```

## 📤 Exporting Data

```python

df.to_csv("cleaned_netflix.csv", index=False)
df.to_excel("cleaned_netflix.xlsx", index=False)
```

```python

# Vectorization instead of loops
df['year_plus_1'] = df['release_year'] + 1
```
